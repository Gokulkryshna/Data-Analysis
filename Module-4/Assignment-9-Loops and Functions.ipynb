Number Guessing Game 

```python

import random

secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
    guess = int(input("Guess the number between 1 and 10: "))
    print("Guess the number between 1 and 10: ", guess)

    if guess < 1 or guess > 10:
        print("Your guess is out of range. Guess a number between 1 and 10.")
        continue

    attempts = attempts - 1

    if guess == secret_number:
        print("Wow! Congratulations, you guessed the correct number!")
        break
    elif guess > secret_number:
        print("Too high, try again.")
    else:
        print("Too low, guess again.")

else:
    print("Your attempts have ended. Try next time!")
    print("The secret number was:", secret_number)


```

Multiplication Table Generator

```python
number =int(input("Enter the number for which you want the multiplication table"))
print("Enter the number for which you want the multiplication table:",number)
ans = 0
for i in range (1,11):
    ans = number * i
    print(number,"x",i," ","=",ans)
```

BMI Calculator

```python
def calculate_bmi(weight,height):
    bmi = weight /(height ** 2)
    return bmi
weight = float(input("Enter your weight in kg"))
print("enter your weight in kg-",weight)
height = float(input("Enter your height in meters"))
print("enter your height in meters-",height)
bmi = calculate_bmi(weight,height)
print("Your BMI is:",format(bmi,".2f"))
```
