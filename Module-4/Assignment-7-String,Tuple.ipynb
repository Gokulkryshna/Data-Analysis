String Concatenation

```python
string1="hello"
string2=str(input('enter a name'))
string3= string1 +' '+string2
print(string3)
string4="Welcome to Python Programming"
string5=  string3 + ','+  " " +string4
print(string5)
```

String Slicing and Indexing

```python
string="Hello Zara, Welcome to Python Programming"

print(string[0])   #first character
print(string[-1])  #last character
print(string[:5])  #first 5 characters
print(string[-11:]) #last 11 characters
print(string[::-1]) #string reverse
print(string[23:29]) #printing python
```

String Methods

```python
strM="Python beginner tutorial"
print(strM.upper())   #uppercase
print(strM.lower())   #lowercase
print(strM.capitalize())   #Orginal input form
print(strM.count('t'))     #count of t
print(strM.replace("Python", "Machine Learning")) #repacing python with ML
```

Tuples

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
t_combine = tuple1 + tuple2
print(t_combine)              #concatenating 2 tuples
print(t_combine * 3)           #repeating 3 times
print(t_combine[2])            #accesing 3rd element
print(t_combine[:3])           # accesing first 3 elements
print(t_combine[-3:])          # accessing last 3 elements
```
