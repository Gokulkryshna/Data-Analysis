List(Creation,Modification,Access)

```python
age_list=[24,25,26,27,28]            #list creation of integers
print(age_list)

name_list=["Zayn","Harry","Louis","Liam","Nial"]
print(name_list)                      #list of string

name_list.append("Yazhini")          #apppend string
print(name_list)

name_list.remove("Yazhini")          #remove string
print(name_list)

age_list.pop(-1)                    #Poped last element
print(age_list)

age_list.extend([29,30,26])        #list extended
print(age_list)

age_list.sort(reverse=True)         #descending Order
print(age_list)

print('Maximum age=',max(age_list))              #maximum age

print('Minimum Age=',min(age_list))              #minimum age

print('Sum of Ages=',sum(age_list))              #sum of all ages

print(name_list[0])                             #first element

print(name_list[-1])                             #Last element

print(name_list[2:5])                            #elemnts from index 2 to 4

print(name_list[::-1])                           #reverse


```

Dictionary (Creation, Modification and Access)

```python
student_marks = {                        #creation of student-mark dict
    "Karthik": 85,
    "Rahul": 76,
    "Zara": 92,
    "Diya": 68,
    "Milan": 88
}
print(student_marks)
print("Rahul mark:", student_marks["Rahul"])           #print the mark of a specific student

student_marks["Soya"] = 80                            #Add a new student
print(student_marks)

student_marks["Rahul"] = 82                          #Mark Updated
print(student_marks)

print("Keys:", student_marks.keys())                  #. Print all keys, values and key-value pairs
print("Values:", student_marks.values())
print("Key-value pairs:", student_marks.items())
```

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}                   #Set Operations
 
print(my_set)

   # my_set[4] = 's'       This produces type error because sets are unordered,so it will not support indexing     

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)               #creation of 2 sets
print("Set 2:", set2) 


print("Union:", set1.union(set2))    #union of sets

print("Intersection:", set1.intersection(set2))    #intersection
```

```python
# conditional statement & Operators
score = int(input("Enter your score (0 to 10): "))
print('Enter your score:',(score))
if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent work! Keep it up!")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
