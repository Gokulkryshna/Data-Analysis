```python
import numpy as np
import pandas  as pd
```

Numpy Arrray Operations

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python

print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
temperatures_f1 = (temperatures_w1 * 9/5) + 32              

print("Temperatures in Fahrenheit:", temperatures_f1)
```

```python
print("Maximum temperature:", np.max(temperatures_w1))
print("Minimum temperature:", np.min(temperatures_w1))
print("Mean temperature:", round(np.mean(temperatures_w1), 2))
```

```python
print("First three days:", temperatures_w1[:3])    # First three days
print("Weekend temperatures:", temperatures_w1[-2:])   # Weekend (last two days)
print("Middle three days:", temperatures_w1[2:5])      # Middle three days
```

```python

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

```python
print("Shape:", temperatures.shape)                # Inspect the 2D array
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)
```

```python
print("Week 1 temperatures:", temperatures[0])   #Extract the temperatures for each week and the weekends (last two days)for both weeks.
print("Week 2 temperatures:", temperatures[1])
print("Week 1 weekend:", temperatures[0, -2:]) 
print("Week 2 weekend:", temperatures[1, -2:])
```

Pandas seriess

```python

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
print("1st rank mark:", marks.iloc[0])    # Access the 1st rank student

print("\nTop 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])     # Retrieve the top 3 ranks 

print("\n3rd rank mark:", marks.iloc[2])   #Retrieve the 3rd rank student

print("\nMarks greater than 90:")     #Filter marks greater than 90 
print(marks[marks > 90])
```

```python
marks.loc['Rank1'] = 100          # Modify the 1st rank student's mark
print("Updated marks:")
print(marks)

marks = marks.drop('Rank5')          # Remove the last rank
print("\nAfter removing Rank5:")
print(marks)

cgpa = marks / 10                    # Calculate CGPA
print("\nCGPA:")
print(cgpa)
```

Pandas DataFrame

```python

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

```python
print("Transactions DataFrame:")
display(transactions)
```

```python
print("First 5 rows:")
display(transactions.head())

```

```python
print("Last 5 rows:")
display(transactions.tail())
```

```python
print("Shape:", transactions.shape)
```

```python
print("Column names:", transactions.columns)
```

```python
print("Data types:")
print(transactions.dtypes)


```

```python
print("Basic information:")
transactions.info()
```

```python
print("Product Category and Amount:")
display(transactions[['ProductCategory', 'Amount']])
```

```python
print("Last 3 columns:")
display(transactions.iloc[:, -3:])
```

```python
print("North region with Amount > 200:")
display(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])
```

```python
print("Product Category Counts:")
print(transactions['ProductCategory'].value_counts())
```

```python
print("Unique Regions:")
print(transactions['Region'].unique())
```

```python
print("Mean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("After modifying Amount:")
display(transactions)
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10
print("After adding Discount:")
display(transactions)
```

```python
transactions = transactions[transactions['TransactionID'] != 109]
print("After removing TransactionID 109:")
display(transactions)
```

```python
transactions = transactions.drop(columns=['Discount'])
print("After deleting Discount column:")
display(transactions)
```
