Data Conversion

```python

df['posted_at'] = pd.to_datetime(df['posted_at'], errors='coerce')
print("Data type:", df['posted_at'].dtype)
```

Data Cleaning

```python

df.isnull().sum()
```

```python

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python

plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False, cmap='viridis')

plt.title("Missing Values in Social Media Dataset")
plt.show()
```

```python
df.isnull().sum()
```

```python
numeric_columns = ['age', 'likes', 'comments', 'shares']

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Numerical missing values filled!")
```

```python
categorical_columns = ['gender', 'sentiment']

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Categorical missing values filled!")
```

```python
df.isnull().sum()
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()
print("Number of Duplicate Rows:", duplicate_count)
```

```python

df['gender'] = df['gender'].str.strip().str.lower()

print(df['gender'].value_counts())

df['sentiment'] = df['sentiment'].str.strip().str.lower()

print(df['sentiment'].value_counts())
```

```python

print("Negative Likes:", (df['likes'] < 0).sum())
print("Negative Comments:", (df['comments'] < 0).sum())
print("Negative Shares:", (df['shares'] < 0).sum())
```

Data Wrangling

```python
# Create engagement score

df['engagement_score'] = (
    df['likes'] + df['comments'] + df['shares']
)

print(df[['likes', 'comments', 'shares', 'engagement_score']].head(10))
```

```python
# Apply log transformation

df['log_likes'] = np.log1p(df['likes'])
df['log_comments'] = np.log1p(df['comments'])
df['log_shares'] = np.log1p(df['shares'])

print(df[['log_likes', 'log_comments', 'log_shares']].head())
```

Statistical Analysis.

```python
# Group data by post type

post_type_analysis = df.groupby('post_type').agg(
    average_likes=('likes', 'mean'),
    average_comments=('comments', 'mean'),
    average_shares=('shares', 'mean'),
    average_engagement=('engagement_score', 'mean')
)

print(post_type_analysis)
```

```python
# Group data by country

country_analysis = df.groupby('country').agg(
    total_posts=('post_id', 'count'),
    average_likes=('likes', 'mean'),
    average_impressions=('impression_count', 'mean'),
    average_engagement_rate=('engagement_rate', 'mean')
)

print(country_analysis)
```

```python
# Group data by sentiment

sentiment_analysis = df.groupby('sentiment').agg(
    total_posts=('post_id', 'count'),
    average_likes=('likes', 'mean'),
    average_comments=('comments', 'mean'),
    average_shares=('shares', 'mean'),
    average_engagement=('engagement_score', 'mean')
)

print(sentiment_analysis)
```

```python
# Sort post types by average engagement

post_type_analysis.sort_values(
    by='average_engagement',
    ascending=False
)
```

```python
# Sort countries by average engagement rate

country_analysis.sort_values(
    by='average_engagement_rate',
    ascending=False
)
```

```python
stats_columns = [
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'engagement_rate',
    'follower_count'
]
```

```python
for column in stats_columns:
    print(f"\nStatistics for {column}")
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().iloc[0])
    print("Standard Deviation:", df[column].std())
    print("Variance:", df[column].var())
    print("25th Percentile:", df[column].quantile(0.25))
    print("50th Percentile:", df[column].quantile(0.50))
    print("75th Percentile:", df[column].quantile(0.75))
```
